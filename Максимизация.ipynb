Нужно максимизировать число A, заменяя в нём цифры на цифры из числа B. При этом каждую цифру из B можно использовать только один раз

```python
def maximize_b(a: str, b: str) -> str:
    a_list = list(a)
    b_list = sorted(b, reverse=True)
    b_index = 0

    for i in range(len(a_list)):
        if b_index < len(b_list) and b_list[b_index] > a_list[i]:
            a_list[i] = b_list[b_index]

    return ''.join(a_list)
```

```python
if __name__ == "__main__":
    a1 = "1234"
    b1 = "5678"
    print(f"test_1('{a1}', '{b1}') = {maximize_b(a1, b1)}")

    a2 = "9876"
    b2 = "5432"
    print(f"test_2('{a2}', '{b2}') = {maximize_b(a2, b2)}")

    # Тест 3
    a3 = "1928"
    b3 = "3476"
    print(f"test_3('{a3}', '{b3}') = {maximize_b(a3, b3)}")

    # Тест 4
    a4 = "2222"
    b4 = "7777"
    print(f"test_4('{a4}', '{b4}') = {maximize_b(a4, b4)}")
```
