Задача — оптимизация ассортимента.

Компания производит два вида продуктов: продукт А и продукт Б.
С помощью модели Курно найди оптимальные значения объема выпуска продукта А и продукта Б для компании

p = 800-q1-q2 – цена товара;

C1 = 7 – себестоимость производства единицы продукта А;
C2 = 2 – себестоимость производства единицы продукта Б.

π1 → max
π2 → max

```python
import numpy as np


A = np.array([[2, 1],
              [1, 2]])

B = np.array([793, 798])
```

```python
solution = np.linalg.solve(A, B)

q1, q2 = solution

print(f"Оптимальный объём выпуска продукта А: {q1:.2f} единиц")
print(f"Оптимальный объём выпуска продукта Б: {q2:.2f} единиц")
```

```python
p = 800 - q1 - q2
profit1 = p * q1 - 7 * q1
profit2 = p * q2 - 2 * q2

print(f"\nЦена на рынке: {p:.2f}")
print(f"Прибыль фирмы 1 (продукт А): {profit1:.2f}")
print(f"Прибыль фирмы 2 (продукт Б): {profit2:.2f}")
```
