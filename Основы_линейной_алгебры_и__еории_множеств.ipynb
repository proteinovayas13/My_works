#ЗДля их доработки используется 2 вида краски: краска для фона (S1) и краска для декора (S2). Нормы расхода краски (на 1 кг.) представлены в виде матрицы:

        24 28

А =  26 21

        22 25

Планируется выпуск новых ящиков каждого вида, указанного в следующем количестве (шт: ):

С = (970 560 1100)



Стоимость единицы каждого типа краски, руб за кг:

        32

В =

       41

Рассчитайте общую стоимость (Q) материалов (краски).

```python
import numpy as np

A = np.array([[24, 28],
              [26, 21],
              [22, 25]])

C = np.array([[970],
              [560],
              [1100]])

B = np.array([[32],
              [41]])
```

```python
# Затраты на каждый вид сырья
A_T = A.T
materials_total = A_T @ C

print(materials_total)
```

```python
# Общая стоимость на 2 вида сырья
costs = materials_total * B
Q = np.sum(costs)
Q
```
